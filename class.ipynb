```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
df=pd.read_excel("B2B_Transaction_Data_.xlsx")
```

```python
df.info()
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df.info()
```

```python
df
```

```python
df["SalesRevenue"] = df["Quantity"] * df["NetPrice"]
```

```python
df
```

```python
df["month"] = pd.DatetimeIndex(df["InvoiceDate"]).month
```

```python
df
```

```python
df_2=df.groupby(["StockCode","month"])["SalesRevenue"].sum().to_frame().reset_index()
```

```python
df_2
```

```python
df_3=df_2.pivot(index="StockCode",columns="month",values="SalesRevenue").fillna(0).reset_index()
```

```python
df_3
```

```python
df_3["total_sales"]= df_3.iloc[:,1:13].sum(axis=1, numeric_only=True) #axis sıfır olunca sütun bazlı topluyor satır satır işlem yapmak demek axis = 1 demek.
```

```python
df_3
```

```python
df_3["Average_sales"]= df_3["total_sales"]/12
df_3["std_dev"]=df_3.iloc[:,1:13].std(axis=1, numeric_only=True)
```

```python
df_3
```

```python
df_3["CV"] = df_3["std_dev"]/df_3["Average_sales"]
```

```python
df_3
```

```python
def xyz_analysis(x):
    if x<=0.5:
        return "X"
    elif x>0.5 and x<=1:
        return "Y"
    else:
        return "Z"
```

```python
df_3["XYZ_Class"]= df_3["CV"].apply(xyz_analysis)
```

```python
df_3
```

```python
df_4=df_3.groupby("StockCode").agg(total_revenue=("total_sales","sum")).sort_values(by="total_revenue",ascending=False).reset_index()
```

```python
df_4
```

```python
df_4["cumulative"]=df_4["total_revenue"].cumsum()
```

```python
df_4
```

```python
df_4["total_cumulative"]=df_4["total_revenue"].sum()
df_4["sku_percent"]=df_4["cumulative"]/df_4["total_cumulative"]
```

```python
df_4
```

```python
def abc_classfication(x):
    if x>0 and x<=0.8:
        return "A"
    elif x>0.8 and x<=0.95:
        return "B"
    else:
        return "C"

df_4["ABC_Class"]= df_4["sku_percent"].apply(abc_classfication)
```

```python
df_4
```

```python
df_3.head()
```

```python
df_3_=df_3[["StockCode","total_sales","Average_sales","std_dev","CV","XYZ_Class"]]
df_4_=df_4[["StockCode","total_revenue","sku_percent","ABC_Class"]]
df_final=pd.merge(df_3_,df_4_,on="StockCode",how="left")
```

```python
df_final    
```

```python
df_final["stock_class"]=df_final["ABC_Class"]+df_final["XYZ_Class"]
```

```python
df_final
```

```python
df_selected_columns=df[["StockCode","Description"]]
df_merge=pd.merge(df_final,df_selected_columns,on="StockCode",how="left")
```

```python
df_result=df_merge.drop_duplicates()
```
